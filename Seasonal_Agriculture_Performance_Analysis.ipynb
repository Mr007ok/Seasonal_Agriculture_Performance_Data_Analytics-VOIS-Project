## 9. Data Cleaning and Preparation



### 9.1 Missing Values

```python
missing = df.isnull().sum().sort_values(ascending=False)
missing_percentage = (df.isnull().mean() * 100).round(2).sort_values(ascending=False)

missing_summary = pd.DataFrame({
    "Missing_Count": missing,
    "Missing_Percentage": missing_percentage
})

missing_summary[missing_summary["Missing_Count"] > 0]
```

### 9.4 Outlier Detection (IQR Method)

 Rather than removing outliers blindly, we first *investigate* them, because in agricultural data an "outlier" may be a genuine, meaningful data point rather than an error.


```python
def iqr_outlier_summary(data, columns):
    rows = []
    for col in columns:
        q1, q3 = data[col].quantile([0.25, 0.75])
        iqr = q3 - q1
        lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
        count = ((data[col] < lower) | (data[col] > upper)).sum()
        rows.append([col, round(lower, 2), round(upper, 2), count, round(count / len(data) * 100, 2)])
    return pd.DataFrame(rows, columns=["Column", "Lower_Bound", "Upper_Bound", "Outlier_Count", "Outlier_%"])

outlier_cols = ["Yield_Tonnes_Ha", "Profit_INR", "Rainfall_mm", "Total_Cost_INR", "Water_Used_m3"]
iqr_outlier_summary(df, outlier_cols)
```

### Q1. How does irrigation method affect profitability, and does the *best* method change by season?

```python
irrigation_profit = df.groupby(["Season", "Irrigation_Method"])["Profit_INR"].mean().round(0).unstack()
irrigation_profit
```

```python
irrigation_profit.plot(kind="bar", figsize=(11, 5))
plt.title("Average Profit by Irrigation Method, per Season")
plt.ylabel("Average Profit (INR)")
plt.xticks(rotation=0)
plt.axhline(0, color="black", linewidth=0.8)
plt.tight_layout()
plt.show()
```

**Answer:** Drip irrigation is the most profitable method in **every** season, and its advantage widens as conditions get harsher: in Zaid, Drip is the *only* irrigation method that still shows a positive average profit — Flood and Rainfed both average a loss. Sprinkler is a consistent middle performer. This suggests Drip irrigation provides the most resilience against seasonal stress.


### Q2. Is there a relationship between disease/pest risk and yield, and does it differ by season?

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5), sharey=True)
for ax, season in zip(axes, df["Season"].unique()):
    subset = df[df["Season"] == season]
    sns.scatterplot(data=subset, x="Disease_Pest_Risk_pct", y="Yield_Tonnes_Ha", alpha=0.4, ax=ax)
    corr = subset["Disease_Pest_Risk_pct"].corr(subset["Yield_Tonnes_Ha"])
    ax.set_title(f"{season} (corr = {corr:.2f})")
plt.tight_layout()
plt.show()
```

**Answer:** Despite Kharif having the *highest average* disease/pest risk, at the individual-farm level the correlation between disease/pest risk and yield is essentially **zero** in all three seasons (correlation between -0.03 and +0.04). In other words, disease/pest risk is a **season-level** pattern (monsoon conditions raise risk broadly) rather than something that predicts which *individual* farms will underperform — farms with high risk scores are not reliably lower-yielding than farms with low risk scores in this dataset.


### Q3. Which state achieves the best water efficiency in each season, and is the leader consistent across seasons?

```python
water_eff_pivot = df.pivot_table(index="State", columns="Season", values="Water_Efficiency_t_per_1000m3", aggfunc="mean").round(2)
water_eff_pivot["Best_Season"] = water_eff_pivot.idxmax(axis=1)
water_eff_pivot
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(water_eff_pivot.drop(columns="Best_Season"), annot=True, fmt=".1f", cmap="Blues")
plt.title("Average Water Efficiency (t per 1000 m³) by State and Season")
plt.tight_layout()
plt.show()
```

**Answer:** The top state changes by season — Gujarat leads in Kharif, Punjab leads in Rabi, and Karnataka leads in Zaid — so there is **no single state that is the best water-efficiency performer year-round** in this dataset. This suggests water efficiency is driven by season-specific regional conditions (e.g., regional rainfall/irrigation infrastructure matched to that season) rather than a fixed state-level advantage.


### Q4. Do higher fertilizer/pesticide inputs actually translate into higher yield?

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

sns.regplot(data=df.sample(1000, random_state=1), x="Fertilizer_kg_ha", y="Yield_Tonnes_Ha",
            scatter_kws={"alpha": 0.3}, line_kws={"color": "red"}, ax=axes[0])
axes[0].set_title(f"Fertilizer vs Yield (corr = {df['Fertilizer_kg_ha'].corr(df['Yield_Tonnes_Ha']):.3f})")

sns.regplot(data=df.sample(1000, random_state=1), x="Pesticide_Litre_ha", y="Yield_Tonnes_Ha",
            scatter_kws={"alpha": 0.3}, line_kws={"color": "red"}, ax=axes[1])
axes[1].set_title(f"Pesticide vs Yield (corr = {df['Pesticide_Litre_ha'].corr(df['Yield_Tonnes_Ha']):.3f})")

plt.tight_layout()
plt.show()
```

**Answer:** No — both fertilizer usage (corr ≈ 0.00) and pesticide usage (corr ≈ -0.01) show **essentially no relationship** with yield in this dataset. The flat regression lines confirm this visually. This is an important, slightly counter-intuitive finding: within the range of inputs recorded here, simply applying more fertilizer or pesticide is not associated with higher yield — crop type and water efficiency are far stronger drivers.


## 18. Key Insights

1. **Kharif is consistently the top-performing season** — highest average yield (5.63 t/ha), highest average profit (₹178,915), and highest water efficiency — while Zaid is consistently the weakest, with a *negative* average profit (-₹24,805).
2. **Profitability declines steadily Kharif → Rabi → Zaid**, and this pattern holds across almost every crop and irrigation method, indicating a genuine seasonal effect rather than a coincidence tied to one crop.
3. **Sugarcane operates on a completely different yield scale** (≈38–53 t/ha) than every other crop (≈0.6–3 t/ha) — any yield analysis must be done *within* crop groups, not on yield alone, or Sugarcane will dominate and distort the picture.
4. **Water efficiency, not rainfall volume, is what actually tracks with yield** — `Water_Efficiency_t_per_1000m3` correlates strongly with yield (≈0.92), while raw `Rainfall_mm` barely correlates with yield at all (≈0.03).
5. **Drip irrigation is the most profitable method in every season**, and is the *only* irrigation method that remains profitable on average during the harsh Zaid season — Flood and Rainfed both average losses in Zaid.
6. **Fertilizer and pesticide use show virtually no relationship with yield** (correlations near zero) — more input is not translating into more output in this dataset, which is a useful (and counter-intuitive) red flag for resource efficiency.
7. **Disease/pest risk is a season-wide pattern, not a farm-level predictor** — Kharif has the highest average risk (54.5%) reflecting monsoon conditions, but within any given season, a farm's individual risk score has almost no correlation with its yield.
8. **The "best" state for water efficiency changes every season** (Gujarat → Kharif, Punjab → Rabi, Karnataka → Zaid) — there is no single regional leader, suggesting local infrastructure/practices are tuned to specific seasonal conditions rather than being uniformly excellent.
9. **Higher-yield crops tend to fetch a lower market price per tonne** (correlation ≈ -0.38 between yield and market price) — a yield-vs-price trade-off exists across the crop portfolio.
10. **Zaid is the smallest season by record count** (~15% of the dataset) — its statistics, while directionally consistent with the other seasons, are based on a notably smaller sample and warrant a little more caution when generalized.


## 19. Final Conclusion

This analysis of 4,000 seasonal farm records confirms that **agricultural performance is strongly and consistently shaped by season**, with Kharif outperforming Rabi, which outperforms Zaid, across yield, profit and water efficiency — a pattern that holds regardless of crop or irrigation method. However, the *drivers* of that performance are more nuanced than season alone: **water efficiency and irrigation method** (particularly Drip irrigation) matter far more than raw rainfall or agronomic input levels (fertilizer, pesticide), and disease/pest risk — while elevated in Kharif — does not predict which individual farms underperform.

These findings translate into concrete, evidence-based guidance: prioritize Drip irrigation (especially in the riskier Zaid season), re-examine input-use efficiency, and design water-management and crop/pricing strategies at the season-and-region level rather than applying a single blanket recommendation. Overall, the project demonstrates that seasonal agricultural data — when properly cleaned, explored and interrogated with targeted questions — yields actionable insight well beyond what is visible in the raw numbers.
